```python
import random
import math


AREA_SIZE = 100
NUM_SENSORS = 5
SENSING_RADIUS = 20
COMM_RADIUS = 40


NUM_PARTICLES = 20
ITERATIONS = 100
W = 0.7
C1 = 1.5
C2 = 1.5


def distance(x1, y1, x2, y2):
    return math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)




def fitness(position):
    """
    Position contains:
    [x1, y1, x2, y2, ..., xn, yn]
    """


    grid_size = 10
    covered = 0
    total_points = 0

    for x in range(0, AREA_SIZE + 1, grid_size):
        for y in range(0, AREA_SIZE + 1, grid_size):

            total_points += 1

            for i in range(NUM_SENSORS):
                sx = position[2 * i]
                sy = position[2 * i + 1]

                if distance(x, y, sx, sy) <= SENSING_RADIUS:
                    covered += 1
                    break

    coverage = covered / total_points


    communication_distance = 0

    for i in range(NUM_SENSORS):
        for j in range(i + 1, NUM_SENSORS):

            d = distance(
                position[2 * i],
                position[2 * i + 1],
                position[2 * j],
                position[2 * j + 1]
            )


            if d > COMM_RADIUS:
                communication_distance += d - COMM_RADIUS



    fitness_value = coverage * 100 - communication_distance * 0.5

    return fitness_value, coverage




def create_particle():
    position = []

    for i in range(NUM_SENSORS):
        position.append(random.uniform(0, AREA_SIZE))
        position.append(random.uniform(0, AREA_SIZE))

    return position



def pso():

    particles = []

    for i in range(NUM_PARTICLES):

        position = create_particle()

        velocity = [
            random.uniform(-5, 5)
            for _ in range(NUM_SENSORS * 2)
        ]

        fit, coverage = fitness(position)

        particles.append({
            "position": position,
            "velocity": velocity,
            "pbest": position[:],
            "pbest_fitness": fit
        })

    gbest = particles[0]["pbest"][:]
    gbest_fitness = particles[0]["pbest_fitness"]

    for particle in particles:

        if particle["pbest_fitness"] > gbest_fitness:
            gbest = particle["pbest"][:]
            gbest_fitness = particle["pbest_fitness"]



    for iteration in range(ITERATIONS):

        for particle in particles:

            position = particle["position"]
            velocity = particle["velocity"]


            for j in range(NUM_SENSORS * 2):

                r1 = random.random()
                r2 = random.random()

                velocity[j] = (
                    W * velocity[j]
                    + C1 * r1 *
                    (particle["pbest"][j] - position[j])
                    + C2 * r2 *
                    (gbest[j] - position[j])
                )


            for j in range(NUM_SENSORS * 2):

                position[j] += velocity[j]


                position[j] = max(
                    0,
                    min(AREA_SIZE, position[j])
                )


            current_fitness, coverage = fitness(position)


            if current_fitness > particle["pbest_fitness"]:

                particle["pbest"] = position[:]
                particle["pbest_fitness"] = current_fitness

            if current_fitness > gbest_fitness:

                gbest = position[:]
                gbest_fitness = current_fitness

        print(
            "Iteration:",
            iteration + 1,
            "Best Fitness:",
            round(gbest_fitness, 2)
        )

    return gbest, gbest_fitness




best_position, best_fitness = pso()

best_fitness_value, coverage = fitness(best_position)

print("\n==============================")
print("OPTIMIZED SENSOR LOCATIONS")
print("==============================")

for i in range(NUM_SENSORS):

    x = best_position[2 * i]
    y = best_position[2 * i + 1]

    print(
        "Sensor", i + 1,
        ": (",
        round(x, 2),
        ",",
        round(y, 2),
        ")"
    )

print("\nCoverage Percentage:",
      round(coverage * 100, 2), "%")

print("Best Fitness:",
      round(best_fitness_value, 2))
```
